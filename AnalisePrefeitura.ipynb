```python
# Em um primeira analise superficial podemos notar que existem muitas colunas do tipo object vamos agora ver todas elas
df.select_dtypes(include='object').columns
```

```python
#vamos verificar como estão salvos os valores para saber como converter
df['Valor Empenhado'].head(20)
```

```python
#Temos uma coluna nomeada de Data mas que está apresentada como um objeto então vamos primeiro converter para uma data
df['Data'] = pd.to_datetime(df['Data'], format='%d/%m/%Y')
```

```python
# Aqui convertemos todas as colunas que estão como objetos para colunas numericas com expressões regulares para remover virgulaas pontos e simbolo de reais
colunas_numericas = [
    'Dotação',
    'Alteração Dotação',
    'Dotação Atual',
    'Valor Anulado',
    'Reforço',
    'Valor Empenhado',
    'Valor Liquidado',
    'Valor Pago',
    'Empenhado até Hoje',
    'Liquidado até Hoje',
    'Pago até Hoje'
]

for coluna in colunas_numericas:
    df[coluna] = (
        df[coluna]
        .astype(str)
        .str.replace('R$', '', regex=False)
        .str.replace('.', '', regex=False)
        .str.replace(',', '.', regex=False)
        .str.strip()
    )

    df[coluna] = pd.to_numeric(df[coluna], errors='coerce')
```

```python
print(df.nunique())
```

```python
df[colunas_numericas].describe()
```

```python
#Agora vamos remover todos os registros que estão completamente vazios
df = df.dropna(how='all').copy()
```

```python
#Feita a remoção vamos verificar os valores ausentes novamente
ausentes = pd.DataFrame({
    'Quantidade': df.isnull().sum(),
    'Percentual': (
        df.isnull().mean() * 100
    ).round(2)
})

ausentes.sort_values(
    by='Quantidade',
    ascending=False
)
```

**Vemos que as colunas que apresentam mais valores ausentes são Processo licitatório que pode indicar que não existe um processo ligado aquela despesa e Nome fonte STN que ainda vamos investigar a existencia de algum código de fonte correspondente Antes de decidir como prosseguir**

```python
# Seguimos com a padronização, agora dos campos de texto, removendo espaços desncessários no inicio e final dos textos
colunas_texto = df.select_dtypes(
    include=['object', 'string']
).columns

for coluna in colunas_texto:

    df[coluna] = (
        df[coluna]
        .astype('string')
        .str.strip()
    )
```

```python
# Agora vamos preparar todos os campos identificadores
colunas_inteiras = [
    'Empenho',
    'N° Ficha',
    'Cód. Forn.',
    'Local',
    'Função',
    'Subfunção',
    'Fonte'
]

for coluna in colunas_inteiras:

    df[coluna] = pd.to_numeric(
        df[coluna],
        errors='coerce'
    ).astype('Int64')
```

```python
# Nas colunas que representam códigos decidi por manter o formato de texto
colunas_codigos = [
    'Funcional',
    'Cód. de aplicação',
    'Natureza',
    'Cód. Fonte',
    'Fonte STN'
]

for coluna in colunas_codigos:

    df[coluna] = df[coluna].astype('string')
```
